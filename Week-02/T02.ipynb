## 2. Concept Explanation

### The Hadamard Gate ($H$) and Superposition
In quantum computing, every newly initialized qubit begins in the default ground computational basis state $|0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$.

The Hadamard gate is a fundamental single-qubit unitary gate with matrix representation:
$$H = \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}$$

Applying $H$ to the ground state transforms it into an equal superposition state $|+\rangle$:
$$H|0\rangle = \frac{|0\rangle + |1\rangle}{\sqrt{2}} = \begin{pmatrix} \frac{1}{\sqrt{2}} \\ \frac{1}{\sqrt{2}} \end{pmatrix} \approx \begin{pmatrix} 0.7071 \\ 0.7071 \end{pmatrix}$$
In this state, measuring in the computational basis yields $|0\rangle$ or $|1\rangle$ with equal probability:
$$P(|0\rangle) = \left|\frac{1}{\sqrt{2}}\right|^2 = 0.5 \quad (50\%), \qquad P(|1\rangle) = \left|\frac{1}{\sqrt{2}}\right|^2 = 0.5 \quad (50\%)$$

### The Phase Gate ($S$)
The Phase gate ($S$, also known as the $P(\pi/2)$ gate) is a single-qubit unitary operation that applies a relative phase shift of $\frac{\pi}{2}$ ($90^\circ$) to the $|1\rangle$ component while keeping the $|0\rangle$ component unchanged:
$$S = \begin{pmatrix} 1 & 0 \\ 0 & e^{i\pi/2} \end{pmatrix} = \begin{pmatrix} 1 & 0 \\ 0 & i \end{pmatrix}$$

Applying $S$ to individual basis states gives:
$$S|0\rangle = |0\rangle, \qquad S|1\rangle = i|1\rangle$$

### Combined Operation & Complex-Valued Statevector
When the $S$ gate acts upon the equal superposition state $|+\rangle = \frac{|0\rangle + |1\rangle}{\sqrt{2}}$, it produces:
$$|\psi\rangle = S|+\rangle = S\left(\frac{|0\rangle + |1\rangle}{\sqrt{2}}\right) = \frac{|0\rangle + i|1\rangle}{\sqrt{2}} = \begin{pmatrix} \frac{1}{\sqrt{2}} \\ \frac{i}{\sqrt{2}} \end{pmatrix} \approx \begin{pmatrix} 0.70710678 + 0j \\ 0 + 0.70710678j \end{pmatrix}$$

This combined transformation creates a **complex-valued statevector** where the $|1\rangle$ amplitude is purely imaginary ($+0.70710678i$).

> **Important Note on Probabilities**: The $S$ gate does **not** change the computational basis measurement probabilities of the $|+\rangle$ state. Since $|i/\sqrt{2}|^2 = |1/\sqrt{2}|^2 = 0.5$, both states continue to have exactly a $50\%$ probability of being measured in the $Z$-basis. The $S$ gate strictly modifies the **relative phase** ($\Delta\phi = +\pi/2$).

### The Bloch Sphere and Geometric Representation
The **Bloch sphere** is a unit 2-sphere representation of a single-qubit pure quantum state. Any single-qubit pure state can be written as:
$$|\psi\rangle = \cos\left(\frac{\theta}{2}\right)|0\rangle + e^{i\phi}\sin\left(\frac{\theta}{2}\right)|1\rangle$$
where:
- $\theta \in [0, \pi]$ is the polar angle (colatitude), determining the relative amplitudes of $|0\rangle$ and $|1\rangle$.
- $\phi \in [0, 2\pi)$ is the azimuthal angle (longitude), representing the relative quantum phase.

### Effect of Relative Phase on Bloch Sphere Coordinates
- **Initial State $|0\rangle$**: Lies at the North Pole ($\theta = 0$).
- **After Hadamard Gate ($|+\rangle$)**: Positioned on the equator along the positive $X$-axis ($\theta = \pi/2, \phi = 0$).
- **After Phase Gate ($|+i\rangle$)**: The $S$ gate induces a $+90^\circ$ ($\pi/2$) rotation around the $Z$-axis. The state vector rotates along the equatorial plane from the positive $X$-axis ($\phi = 0$) directly to the positive $Y$-axis ($\phi = \pi/2$), corresponding to the state $|+i\rangle = \frac{|0\rangle + i|1\rangle}{\sqrt{2}}$.

## 3. Circuit Flow
The transformation flow of this experiment is:
$$|0\rangle \xrightarrow{\quad \text{Hadamard } (H) \quad} |+\rangle = \frac{|0\rangle + |1\rangle}{\sqrt{2}} \xrightarrow{\quad \text{Phase } (S) \quad} |+i\rangle = \frac{|0\rangle + i|1\rangle}{\sqrt{2}}$$

1. **State Preparation**: Qubit $q_0$ is initialized in $|0\rangle$.
2. **Superposition Generation**: Hadamard gate $H$ rotates the state to $|+\rangle$ along the $+X$ axis.
3. **Phase Rotation**: Phase gate $S$ rotates the state by $\pi/2$ around the $Z$-axis into $|+i\rangle$ along the $+Y$ axis.
4. **Statevector & Visualization**: The resulting statevector is computed and projected onto the Bloch sphere.

## 4. Tools and Libraries Used
- **Qiskit**: IBM's open-source quantum computing software development kit (`QuantumCircuit`).
- **qiskit.quantum_info.Statevector**: Class used for exact mathematical extraction and direct evaluation of the quantum statevector without requiring measurement collapse or shot-based sampling.
- **Qiskit Visualization Utilities (`plot_bloch_multivector`)**: Special function for rendering quantum states as 3D vectors on the Bloch sphere.
- **Matplotlib (`matplotlib.pyplot`)**: Plotting backend for displaying the Bloch multivector visualization.

## Implementation

### Import Required Libraries
Import `QuantumCircuit` and `Statevector` from Qiskit, along with `plot_bloch_multivector` and `matplotlib.pyplot` for Bloch sphere visualization.

### Create the Single-Qubit Circuit
Initialize a 1-qubit quantum circuit (`QuantumCircuit(1)`) in the default $|0\rangle$ state.

### Apply the Hadamard Gate
Apply `qc.h(0)` to place the qubit into an equal superposition state $|+\rangle$.

### Apply the Phase (S) Gate
Apply `qc.s(0)` to introduce a relative phase of $\pi/2$ to the $|1\rangle$ component.

### Obtain the Statevector
Extract the complex statevector using `Statevector.from_instruction(qc)`.

### Visualize the State on the Bloch Sphere
Display the circuit diagram, print the numerical statevector, and visualize the quantum state on the Bloch sphere using `plot_bloch_multivector`.

```python
# Import Required Libraries
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt

# Create the Single-Qubit Circuit
qc = QuantumCircuit(1)

# Apply the Hadamard Gate
qc.h(0)

# Apply the Phase (S) Gate
qc.s(0)

# Display the Quantum Circuit
print("Quantum Circuit:\n")
print(qc.draw())

# Obtain the Statevector
state = Statevector.from_instruction(qc)

# Display the Statevector
print("\nStatevector:")
print(state)

# Visualize the State on the Bloch Sphere
plot_bloch_multivector(state)
plt.show()
```

## Conclusion

This experiment demonstrates a two-step single-qubit quantum transformation and analyzes its geometrical representation:

1. **Superposition and Relative Phase Transformation**:
   - The **Hadamard gate ($H$)** created an equal superposition of basis states:
     $$|0\rangle \longrightarrow |+\rangle = \frac{|0\rangle + |1\rangle}{\sqrt{2}}$$
   - The **Phase gate ($S$)** introduced a relative phase of $\frac{\pi}{2}$ ($90^\circ$) to the $|1\rangle$ component without affecting the $|0\rangle$ component:
     $$|+\rangle \longrightarrow |+i\rangle = \frac{|0\rangle + i|1\rangle}{\sqrt{2}}$$

2. **Statevector Verification**:
   - The resulting statevector extracted directly from the circuit is:
     $$\text{Statevector}([0.70710678+0.j, \; 0.+0.70710678j], \; \text{dims}=(2,))$$
   - The amplitude of $|0\rangle$ is $\frac{1}{\sqrt{2}} \approx 0.70710678$, and the amplitude of $|1\rangle$ is $\frac{i}{\sqrt{2}} \approx 0.70710678i$.
   - The $S$ gate strictly altered the **relative phase** ($\Delta\phi = \pi/2$) while leaving computational basis measurement probabilities invariant ($P(|0\rangle) = 0.5$, $P(|1\rangle) = 0.5$).

3. **Bloch-Sphere Representation**:
   - The state $|+\rangle$ points along the positive $X$-axis of the Bloch sphere.
   - The phase shift from the $S$ gate corresponds geometrically to a $+90^\circ$ rotation around the $Z$-axis, rotating the state vector from the $+X$ axis to the **positive $Y$-axis** ($|y+\rangle = |+i\rangle$).
   - The statevector and Bloch-sphere representation together provide an intuitive and rigorous way to observe and verify relative phase transformations in quantum computing.
