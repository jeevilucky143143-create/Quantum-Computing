## 2. Concept Explanation

### Computational Basis State Initialization
In quantum computing, a two-level quantum system (qubit) is described in a two-dimensional complex Hilbert space $\mathcal{H}_2$. The standard computational basis consists of the two orthonormal states $|0\rangle$ and $|1\rangle$:
$$|0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \quad |1\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$$

By convention in quantum computing and Qiskit, every newly allocated qubit is initialized into the ground computational basis state $|0\rangle$.

### The Pauli-X Gate (Quantum Bit-Flip)
The Pauli-X gate ($X$) is the quantum analogue of the classical NOT gate (bit-flip operation). It is represented by the Pauli matrix $\sigma_x$. Geometrically on the Bloch sphere, it corresponds to a $\pi$-radian ($180^\circ$) rotation around the $X$-axis.

Its unitary matrix representation in the computational basis is:
$$X = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}$$

Applying the Pauli-X operator to the computational basis states yields:
$$X|0\rangle = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix} \begin{pmatrix} 1 \\ 0 \end{pmatrix} = \begin{pmatrix} 0 \\ 1 \end{pmatrix} = |1\rangle$$
$$X|1\rangle = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix} \begin{pmatrix} 0 \\ 1 \end{pmatrix} = \begin{pmatrix} 1 \\ 0 \end{pmatrix} = |0\rangle$$

Hence, the Pauli-X gate inverts the computational state, mapping $|0\rangle \mapsto |1\rangle$ and $|1\rangle \mapsto |0\rangle$.

### Statevector Representation in Quantum Computing
A **statevector** is a normalized complex vector that encapsulates the complete quantum state of a closed quantum system. For a single qubit, an arbitrary pure quantum state $|\psi\rangle$ is represented as a linear combination of the computational basis states:
$$|\psi\rangle = \alpha |0\rangle + \beta |1\rangle = \begin{pmatrix} \alpha \\ \beta \end{pmatrix}, \quad \alpha, \beta \in \mathbb{C}$$
subject to the normalization condition:
$$\langle \psi | \psi \rangle = |\alpha|^2 + |\beta|^2 = 1$$
Here, $\alpha$ and $\beta$ are complex probability amplitudes, where $|\alpha|^2$ and $|\beta|^2$ represent the measurement probabilities for states $|0\rangle$ and $|1\rangle$, respectively.

### Resulting Statevector for the Circuit
Starting from the ground state $|0\rangle$ and applying the Pauli-X gate produces the state:
$$|\psi\rangle = X|0\rangle = 0|0\rangle + 1|1\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$$
The resulting statevector directly represents the $|1\rangle$ computational basis state, with zero probability amplitude for $|0\rangle$ and unit probability amplitude for $|1\rangle$.

## 3. Circuit Flow
The sequence of quantum operations is:
$$|0\rangle \xrightarrow{\quad \text{Pauli-}X\text{ Gate} \quad} |1\rangle$$

1. **Initialization**: The qubit $q_0$ is prepared in the default ground state $|0\rangle$.
2. **Transformation**: The Pauli-X gate is applied to $q_0$, performing a deterministic bit-flip ($X|0\rangle = |1\rangle$).
3. **Statevector Inspection**: The resulting statevector is extracted and displayed to verify the $|1\rangle$ state without projective measurement collapse.

- **qiskit.quantum_info.Statevector**: Module for direct analytical computation and inspection of exact quantum statevectors without statistical sampling or measurement shots.

## Implementation

### Import Required Libraries
Import `QuantumCircuit` and `Statevector` from Qiskit, along with `plot_bloch_multivector` and `matplotlib.pyplot` for visualization.

### Create the Quantum Circuit
Initialize a 1-qubit quantum circuit (`QuantumCircuit(1)`). By default, the qubit starts in state $|0\rangle$.

### Apply the Pauli-X Gate
Apply the Pauli-X gate (`qc.x(0)`) to perform the quantum bit-flip operation.

### Obtain the Statevector
Compute and extract the exact quantum statevector from the circuit using `Statevector.from_instruction(qc)`.

### Display the Circuit and Statevector
Print the ASCII circuit diagram, display the resulting statevector representation, and render the Bloch sphere multivector.

```python
# Import Required Libraries
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt

# Create the Quantum Circuit
qc = QuantumCircuit(1)

# Apply the Pauli-X Gate
qc.x(0)

# Display the Quantum Circuit
print("Quantum Circuit:\n")
print(qc.draw())

# Obtain the Statevector
state = Statevector.from_instruction(qc)

# Display the Statevector and Bloch Sphere Visualization
print("\nStatevector:")
print(state)

plot_bloch_multivector(state)
plt.show()
```
