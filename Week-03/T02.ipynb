### 2. Concept Explanation

#### Initial State
The single-qubit quantum system begins initialized in the standard computational basis ground state:
$$|0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$$

#### Superposition via the Hadamard Gate
The single-qubit Hadamard gate ($H$) transforms computational basis states into symmetric equal-superposition states:
$$H = \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}$$
Applying the Hadamard gate to the initial ground state $|0\rangle$ produces the $|+\rangle$ state:
$$H|0\rangle = \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix} \begin{pmatrix} 1 \\ 0 \end{pmatrix} = \frac{|0\rangle + |1\rangle}{\sqrt{2}} = |+\rangle = \begin{pmatrix} \frac{1}{\sqrt{2}} \\ \frac{1}{\sqrt{2}} \end{pmatrix}$$

#### Relative Phase Flip via the Pauli-Z Gate
The Pauli-Z gate acts as a phase-flip operator:
$$Z = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}$$
It acts on the computational basis states as:
$$Z|0\rangle = |0\rangle, \quad Z|1\rangle = -|1\rangle$$
The Pauli-Z gate leaves $|0\rangle$ invariant and inverts the sign (imparts a relative phase shift of $\pi$ radians) of the $|1\rangle$ component.

#### Complete Transformation: Constructing the $|-\rangle$ State
When the Pauli-Z gate is applied to the equal superposition state $|+\rangle = H|0\rangle$, the relative phase of $|1\rangle$ is inverted:
$$ZH|0\rangle = Z\left(\frac{|0\rangle + |1\rangle}{\sqrt{2}}\right) = \frac{Z|0\rangle + Z|1\rangle}{\sqrt{2}} = \frac{|0\rangle - |1\rangle}{\sqrt{2}} = |-\rangle$$
In matrix form:
$$ZH = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix} \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix} = \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & 1 \\ -1 & 1 \end{pmatrix}$$
$$ZH|0\rangle = \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & 1 \\ -1 & 1 \end{pmatrix} \begin{pmatrix} 1 \\ 0 \end{pmatrix} = \begin{pmatrix} \frac{1}{\sqrt{2}} \\ -\frac{1}{\sqrt{2}} \end{pmatrix} = |-\rangle$$

#### Theoretical Statevector
The analytical statevector representation of the $|-\rangle$ state is:
$$|-\rangle = \begin{pmatrix} \frac{1}{\sqrt{2}} \\ -\frac{1}{\sqrt{2}} \end{pmatrix} \approx \begin{pmatrix} 0.70710678 \\ -0.70710678 \end{pmatrix}$$

#### Numerical Statevector Verification
In numerical computing, floating-point arithmetic introduces tiny round-off errors. Rather than strict equality (`==`), verification compares the theoretical and computed statevectors element-by-element using NumPy's `np.allclose()`, checking that:
$$|a_i - b_i| \le (\text{atol} + \text{rtol} \cdot |b_i|)$$
for all components within floating-point tolerance.

### 3. Circuit Flow
The sequence of quantum operations is:
$$|0\rangle \xrightarrow{\quad \text{Hadamard }(H) \quad} |+\rangle = \frac{|0\rangle + |1\rangle}{\sqrt{2}} \xrightarrow{\quad \text{Pauli-}Z (Z) \quad} |-\rangle = \frac{|0\rangle - |1\rangle}{\sqrt{2}}$$

1. **Initialization**: The single qubit starts in the ground state $|0\rangle$.
2. **Hadamard ($H$)**: Creates an equal superposition with equal relative phase ($|+\rangle$).
3. **Pauli-Z ($Z$)**: Inverts the phase of $|1\rangle$, producing the $|-\rangle$ state.
4. **Statevector Inspection**: The resulting statevector is extracted and compared against the theoretical prediction.

```python
import random
import numpy as np

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator
```

### Statevector Inspection Helper Function
Define a reusable helper function `show_state(circuit, title)` that extracts the exact statevector using `Statevector.from_instruction()`, prints the circuit diagram, and outputs the complex statevector amplitudes.

```python
def show_state(circuit, title):
    state = Statevector.from_instruction(circuit)
    print("="*60)
    print(title)
    print(circuit.draw())
    print("\nStatevector:")
    print(state)
    print("="*60)
    return state
```

## Create the |-> State

Construct a single-qubit quantum circuit initialized in $|0\rangle$, apply a Hadamard gate to create $|+\rangle$, and follow it with a Pauli-Z gate to produce the $|-\rangle$ state.

### Obtain the Statevector
Extract and display the circuit diagram and the resulting quantum statevector using `show_state()`.

```python
qc_minus = QuantumCircuit(1)

qc_minus.h(0)
qc_minus.z(0)

state_minus = show_state(qc_minus, "|-> State using H and Z")
```

---

## Define the Expected Statevector and Verify the Result

### Define the Expected Statevector
Define the theoretical statevector $[1/\sqrt{2}, -1/\sqrt{2}]$ as a NumPy array:
$$\text{expected} = \begin{pmatrix} \frac{1}{\sqrt{2}} \\ -\frac{1}{\sqrt{2}} \end{pmatrix} \approx \begin{pmatrix} 0.70710678 \\ -0.70710678 \end{pmatrix}$$

### Verify the Result
Compare the theoretical statevector against the simulated quantum statevector (`state_minus.data`) using `np.allclose()`, verifying agreement within floating-point numerical tolerance.

```python
expected = np.array([1/np.sqrt(2), -1/np.sqrt(2)])

print("Expected State:")
print(expected)

print("\nObtained State:")
print(state_minus.data)

print("\nVerification:")
print(np.allclose(expected, state_minus.data))
```

- The subsequent **Pauli-Z ($Z$) gate** leaves the $|0\rangle$ component unaltered while flipping the phase of the $|1\rangle$ component ($Z|1\rangle = -|1\rangle$).
- The combination of operations yields the **$|-\rangle$ state**, corresponding to the analytical statevector $[1/\sqrt{2}, -1/\sqrt{2}] \approx [0.70710678, -0.70710678]$.
- Verification using `np.allclose()` confirms exact element-by-element agreement between the computed statevector and theoretical prediction within floating-point numerical precision.
