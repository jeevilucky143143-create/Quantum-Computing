### 2. Concept Explanation

#### Fundamental Single-Qubit Quantum Gates
The Pauli operators ($\sigma_x, \sigma_y, \sigma_z$, or $X, Y, Z$) form the fundamental set of single-qubit quantum gates. They are both unitary ($U^\dagger U = I$) and Hermitian ($U^\dagger = U$), meaning they are self-inverse operators:
$$X^2 = Y^2 = Z^2 = I$$
In quantum information, the Pauli matrices serve as an orthogonal basis for the space of $2 \times 2$ Hermitian matrices and generate $\pi$-radian ($180^\circ$) rotations around the principal axes of the Bloch sphere.

The unitary matrix representations in the standard computational basis $\{|0\rangle, |1\rangle\}$ are:
$$X = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}, \quad Y = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix}, \quad Z = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}$$

#### Pauli-X Gate (Bit-Flip)
The Pauli-X gate acts as the quantum analogue of the classical NOT gate, mapping $|0\rangle$ to $|1\rangle$ and $|1\rangle$ to $|0\rangle$:
$$X|0\rangle = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix} \begin{pmatrix} 1 \\ 0 \end{pmatrix} = \begin{pmatrix} 0 \\ 1 \end{pmatrix} = |1\rangle$$
$$X|1\rangle = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix} \begin{pmatrix} 0 \\ 1 \end{pmatrix} = \begin{pmatrix} 1 \\ 0 \end{pmatrix} = |0\rangle$$
Applying $X$ to the initial ground state $|0\rangle$ performs a deterministic bit-flip, yielding state $|1\rangle$.

#### Pauli-Y Gate (Bit-and-Phase Flip with Complex Phase)
The Pauli-Y gate performs a combined bit-flip and phase-flip, introducing an imaginary unit phase factor $i$:
$$Y|0\rangle = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix} \begin{pmatrix} 1 \\ 0 \end{pmatrix} = \begin{pmatrix} 0 \\ i \end{pmatrix} = i|1\rangle$$
$$Y|1\rangle = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix} \begin{pmatrix} 0 \\ 1 \end{pmatrix} = \begin{pmatrix} -i \\ 0 \end{pmatrix} = -i|0\rangle$$
When applied to $|0\rangle$, it flips the state to $|1\rangle$ with an imaginary amplitude of $+i = e^{i\pi/2}$.

#### Pauli-Z Gate (Phase-Flip)
The Pauli-Z gate acts as a phase-flip operator:
$$Z|0\rangle = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix} \begin{pmatrix} 1 \\ 0 \end{pmatrix} = \begin{pmatrix} 1 \\ 0 \end{pmatrix} = |0\rangle$$
$$Z|1\rangle = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix} \begin{pmatrix} 0 \\ 1 \end{pmatrix} = \begin{pmatrix} 0 \\ -1 \end{pmatrix} = -|1\rangle$$
The Pauli-Z gate leaves $|0\rangle$ unchanged ($Z|0\rangle = |0\rangle$) because $|0\rangle$ is an eigenstate of $Z$ with eigenvalue $+1$. While the Z gate changes the phase of $|1\rangle$ by flipping its sign ($Z|1\rangle = -|1\rangle$), its action on the initial $|0\rangle$ state leaves the state completely unchanged.

#### Statevector Representation
A **statevector** is a normalized complex column vector that stores the exact complex probability amplitudes of the computational-basis states:
$$|\psi\rangle = \alpha |0\rangle + \beta |1\rangle = \begin{pmatrix} \alpha \\ \beta \end{pmatrix}, \quad \alpha, \beta \in \mathbb{C}$$
subject to normalization $\langle\psi|\psi\rangle = |\alpha|^2 + |\beta|^2 = 1$. The statevector representation provides direct access to the exact complex amplitudes and phase angles without measurement collapse or projective sampling noise.

### Statevector Inspection Helper Function
Define a reusable helper function `show_state(circuit, title)` that extracts the exact statevector from the circuit using `Statevector.from_instruction()`, prints the ASCII circuit diagram, and displays the statevector.

```python
def show_state(circuit, title):
    state = Statevector.from_instruction(circuit)
    print("="*60)
    print(title)
    print(circuit.draw())
    print("\nStatevector:")
    print(state)
    print("="*60)
    return state
```

## Pauli-X Gate

Construct a 1-qubit circuit initialized in $|0\rangle$, apply the Pauli-X gate via `qc_x.x(0)`, and inspect the resulting statevector.

The Pauli-X gate performs a bit-flip on $|0\rangle$:
$$X|0\rangle = |1\rangle \implies \begin{pmatrix} 0 \\ 1 \end{pmatrix}$$

```python
qc_x = QuantumCircuit(1)
qc_x.x(0)

state_x = show_state(qc_x, "X Gate")
```

## Pauli-Y Gate

Construct a 1-qubit circuit initialized in $|0\rangle$, apply the Pauli-Y gate via `qc_y.y(0)`, and inspect the resulting statevector.

The Pauli-Y gate performs a combined bit-flip and phase-shift, introducing an imaginary unit factor $i$:
$$Y|0\rangle = i|1\rangle \implies \begin{pmatrix} 0 \\ i \end{pmatrix}$$

```python
qc_y = QuantumCircuit(1)
qc_y.y(0)

state_y = show_state(qc_y, "Y Gate")
```

## Pauli-Z Gate

Construct a 1-qubit circuit initialized in $|0\rangle$, apply the Pauli-Z gate via `qc_z.z(0)`, and inspect the resulting statevector.

The Pauli-Z gate acts as a phase-flip operator. Because $|0\rangle$ is an eigenstate of $Z$ with eigenvalue $+1$, the state is left unchanged:
$$Z|0\rangle = |0\rangle \implies \begin{pmatrix} 1 \\ 0 \end{pmatrix}$$
Although the Z gate changes the phase of $|1\rangle$ ($Z|1\rangle = -|1\rangle$), its action on the initial $|0\rangle$ ground state leaves the statevector unaltered.

```python
qc_z = QuantumCircuit(1)
qc_z.z(0)

state_z = show_state(qc_z, "Z Gate")
```

## Statevector Interpretation

### Summary of State Transformations
The table below compares the theoretical gate operations with the observed statevectors extracted from the circuits:

| Gate | Unitary Matrix | Transformation on $|0\rangle$ | Observed Statevector $[\alpha_0, \alpha_1]$ | Computational Basis Decomposition |
| :---: | :---: | :---: | :---: | :---: |
| **$X$** | $\begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}$ | $|0\rangle \mapsto |1\rangle$ | `[0.+0.j, 1.+0.j]` | $0|0\rangle + 1|1\rangle = |1\rangle$ |
| **$Y$** | $\begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix}$ | $|0\rangle \mapsto i|1\rangle$ | `[0.+0.j, 0.+1.j]` | $0|0\rangle + i|1\rangle = i|1\rangle$ |
| **$Z$** | $\begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}$ | $|0\rangle \mapsto |0\rangle$ | `[1.+0.j, 0.+0.j]` | $1|0\rangle + 0|1\rangle = |0\rangle$ |

### Analytical Insights
1. **Direct Amplitude and Phase Observation**:
   A statevector directly captures both the magnitude and complex phase of each basis amplitude. In particular, the Pauli-Y transformation yields an amplitude of `0.+1.j` ($+i$) for the $|1\rangle$ basis state. Classical projective measurements would only measure the probability magnitude $|\alpha_1|^2 = |i|^2 = 1.0$, completely discarding the complex phase factor $i$. Direct statevector analysis preserves this crucial phase information.
2. **Eigenstate Invariance under $Z$**:
   The Pauli-Z gate acts non-trivially on $|1\rangle$ by flipping its sign ($Z|1\rangle = -|1\rangle$), but leaves $|0\rangle$ unchanged ($Z|0\rangle = |0\rangle$). Because the initial qubit state is $|0\rangle$, the output statevector remains `[1.+0.j, 0.+0.j]`.
3. **Exact Statevector vs. Measurement Sampling**:
   This experiment does not involve measurement sampling, shots, or statistical randomness. By using `Statevector.from_instruction()`, the exact unitary matrix operations are applied deterministically to the initial state, providing an exact mathematical verification of single-qubit Pauli gate transformations.
