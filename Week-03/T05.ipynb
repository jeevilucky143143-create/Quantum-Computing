## Quantum Gate Identity

### Single-Qubit Gate Definitions
- **Hadamard Gate ($H$)**: Creates an equal superposition between the computational basis states $|0\rangle$ and $|1\rangle$, mapping the $Z$-basis into the $X$-basis and vice versa.
- **Pauli-Z Gate ($Z$)**: Performs a phase-flip operation, leaving $|0\rangle$ unchanged and flipping the phase of $|1\rangle$ ($Z|0\rangle = |0\rangle, Z|1\rangle = -|1\rangle$).
- **Pauli-X Gate ($X$)**: Performs a bit-flip operation, interchanging $|0\rangle$ and $|1\rangle$ ($X|0\rangle = |1\rangle, X|1\rangle = |0\rangle$).

### The Identity: $HZH = X$
In quantum computing, the Hadamard transformation acts as a basis-change operator that conjugates the Pauli-Z operator into the Pauli-X operator:
$$HZH = X$$
Because the Hadamard gate is self-inverse ($H = H^\dagger = H^{-1}$), this identity equivalently demonstrates that $X$ and $Z$ are unitarily similar under Hadamard conjugation: $HXH = Z$ and $HZH = X$.

### Dual Verification Strategy
The identity is verified across two complementary levels:
1. **Circuit-Level Statevector Verification**: Tests whether applying the sequence $H \to Z \to H$ to the initial ground state $|0\rangle$ produces the identical output statevector as applying the single gate $X$ to $|0\rangle$:
   $$HZH|0\rangle = X|0\rangle = |1\rangle$$
2. **Matrix-Level Operator Verification**: Demonstrates operator equality across the entire 2-dimensional Hilbert space by verifying that the matrix product $H @ Z @ H$ equals the matrix $X$ element-by-element for any arbitrary quantum state $|\psi\rangle = \alpha|0\rangle + \beta|1\rangle$.

---

### Circuit-Level Verification Overview
We construct two separate 1-qubit circuits initialized in $|0\rangle$:
- **Circuit 1 ($HZH$)**: $|0\rangle \xrightarrow{\quad H \quad} |+\rangle \xrightarrow{\quad Z \quad} |-\rangle \xrightarrow{\quad H \quad} |1\rangle$
- **Circuit 2 ($X$)**: $|0\rangle \xrightarrow{\quad X \quad} |1\rangle$

The resulting statevectors are extracted using `Statevector.from_instruction()` and verified for numerical equality using NumPy's `np.allclose()`.

---

### Matrix-Level Verification Overview
The standard unitary matrix representations in the computational basis $\{|0\rangle, |1\rangle\}$ are:
$$H = \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}, \quad Z = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}, \quad X = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}$$

Evaluating the matrix product:
$$HZH = H @ Z @ H = \left[ \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix} \right] \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix} \left[ \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix} \right] = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix} = X$$
The product matrix is compared against $X$ using `np.allclose()`.

### Statevector Inspection Helper Function
Define a reusable helper function `show_state(circuit, title)` to display the ASCII circuit diagram and extract the statevector.

```python
def show_state(circuit, title):
    state = Statevector.from_instruction(circuit)
    print("="*60)
    print(title)
    print(circuit.draw())
    print("\nStatevector:")
    print(state)
    print("="*60)
    return state
```

## Circuit Verification

### Construct the HZH Circuit
Initialize a 1-qubit quantum circuit (`QuantumCircuit(1)`) in state $|0\rangle$ and sequentially apply the gate sequence:
$$|0\rangle \xrightarrow{\quad H \quad} |+\rangle \xrightarrow{\quad Z \quad} |-\rangle \xrightarrow{\quad H \quad} |1\rangle$$

```python
qc_hzh = QuantumCircuit(1)

qc_hzh.h(0)

qc_hzh.z(0)

qc_hzh.h(0)

state_hzh = show_state(qc_hzh,"HZH Circuit")
```

### Construct the Pauli-X Circuit
Initialize a 1-qubit quantum circuit (`QuantumCircuit(1)`) in state $|0\rangle$ and apply a single Pauli-X gate:
$$|0\rangle \xrightarrow{\quad X \quad} |1\rangle$$

```python
qc_x2 = QuantumCircuit(1)

qc_x2.x(0)

state_x2 = show_state(qc_x2,"X Gate Circuit")
```

---

## Compare the Statevectors

Compare the statevector obtained from the $HZH$ circuit (`state_hzh.data`) with that from the Pauli-X circuit (`state_x2.data`) using `np.allclose()` to confirm numerical equivalence.

```python
print("HZH State:")
print(state_hzh.data)

print("\nX Gate State:")
print(state_x2.data)

print("\nEquivalent?")

print(np.allclose(state_hzh.data,state_x2.data))
```

## Matrix Representation

Define the analytical matrix representations of the Hadamard ($H$), Pauli-Z ($Z$), and Pauli-X ($X$) operators in the computational basis:
$$H = \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}, \quad Z = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}, \quad X = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}$$

---

## Verify HZH = X Algebraically

Compute the matrix product $HZH = H @ Z @ H$ using NumPy's matrix multiplication operator `@` and verify that $H @ Z @ H = X$ using `np.allclose()`.

```python
print("Analytical Proof")

print()

print("Hadamard Matrix:")

print(np.array([[1,1],[1,-1]])/np.sqrt(2))

print()

print("Z Matrix:")

print(np.array([[1,0],[0,-1]]))

print()

print("X Matrix:")

print(np.array([[0,1],[1,0]]))

print()

H = np.array([[1,1],[1,-1]])/np.sqrt(2)

Z = np.array([[1,0],[0,-1]])

HZH = H @ Z @ H

print("HZH Matrix:")

print(HZH)

print()

print("Is HZH equal to X?")

print(np.allclose(HZH,np.array([[0,1],[1,0]])))
```
