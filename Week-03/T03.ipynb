### 2. Concept Explanation

#### Initial Quantum State
The quantum circuit begins initialized in the standard computational basis ground state:
$$|0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$$

#### Gate Transformation Roles
Each gate in the set $\{X, Y, Z, H\}$ transforms the qubit state through a specific unitary operation:
- **Pauli-X Gate ($X$)**: Performs a bit-flip operation, interchanging $|0\rangle$ and $|1\rangle$:
  $$X = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}, \quad X|0\rangle = |1\rangle, \quad X|1\rangle = |0\rangle$$
- **Pauli-Y Gate ($Y$)**: Performs a combined bit-flip and phase-flip, introducing an imaginary unit phase factor:
  $$Y = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix}, \quad Y|0\rangle = i|1\rangle, \quad Y|1\rangle = -i|0\rangle$$
- **Pauli-Z Gate ($Z$)**: Performs a phase-flip operation, leaving $|0\rangle$ unchanged while flipping the sign of $|1\rangle$:
  $$Z = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}, \quad Z|0\rangle = |0\rangle, \quad Z|1\rangle = -|1\rangle$$
- **Hadamard Gate ($H$)**: Creates an equal superposition between the computational basis states:
  $$H = \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}, \quad H|0\rangle = \frac{|0\rangle + |1\rangle}{\sqrt{2}} = |+\rangle, \quad H|1\rangle = \frac{|0\rangle - |1\rangle}{\sqrt{2}} = |-\rangle$$

#### Final State Dependency
Because single-qubit quantum operations do not in general commute (e.g., $XZ = -ZX \ne ZX$), the overall unitary operation:
$$U_{\text{total}} = U_5 U_4 U_3 U_2 U_1$$
and the resulting final state:
$$|\psi_{\text{final}}\rangle = U_{\text{total}}|0\rangle$$
depend directly on the specific sequence and order of the randomly selected gates.

#### Statevector Representation
A **statevector** represents the complex probability amplitudes of the qubit's computational basis states:
$$|\psi\rangle = \alpha |0\rangle + \beta |1\rangle = \begin{pmatrix} \alpha \\ \beta \end{pmatrix}, \quad \alpha, \beta \in \mathbb{C}, \quad |\alpha|^2 + |\beta|^2 = 1$$
where $|\alpha|^2$ and $|\beta|^2$ denote the probabilities of measuring states $|0\rangle$ and $|1\rangle$, respectively.

```python
import random
import numpy as np

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator
```

### Statevector Inspection Helper Function
Define a reusable helper function `show_state(circuit, title)` that extracts the exact statevector using `Statevector.from_instruction()`, prints the circuit diagram, and displays the statevector.

```python
def show_state(circuit, title):
    state = Statevector.from_instruction(circuit)
    print("="*60)
    print(title)
    print(circuit.draw())
    print("\nStatevector:")
    print(state)
    print("="*60)
    return state
```

---

## Generate a Random Five-Gate Sequence

Define the candidate gate set `['x', 'y', 'z', 'h']` and randomly select exactly five gates using `random.choice()`.

```python
gate_list = ['x','y','z','h']

random_gates = [random.choice(gate_list) for _ in range(5)]

print("Random Gates Selected:\n")
print(random_gates)
```

---

## Construct the Random Quantum Circuit

Initialize a single-qubit quantum circuit (`QuantumCircuit(1)`) in the default $|0\rangle$ ground state. Iterate through the randomly selected sequence and apply each gate to qubit `0`. Display the completed circuit diagram and resulting statevector.

```python
qc_random = QuantumCircuit(1)

for gate in random_gates:

    if gate == 'x':
        qc_random.x(0)

    elif gate == 'y':
        qc_random.y(0)

    elif gate == 'z':
        qc_random.z(0)

    elif gate == 'h':
        qc_random.h(0)

state_random = show_state(qc_random, "Random Five-Gate Circuit")
```

---

## Obtain the Analytical Statevector

Extract and display the complex amplitude data array from the statevector object (`state_random.data`).

```python
print("Analytical Final Statevector:\n")
print(state_random.data)
```

---

## Verify Using AerSimulator

Simulate the random five-gate quantum circuit using Qiskit Aer's `AerSimulator(method="statevector")`.

### Compare the Statevectors
Compare the analytical statevector (`state_random.data`) with the simulated statevector (`state_sim.data`) using `np.allclose()` to verify numerical agreement within floating-point precision.

```python
simulator = AerSimulator(method="statevector")

result = simulator.run(qc_random).result()

state_sim = Statevector.from_instruction(qc_random)

print("Simulation Statevector:\n")
print(state_sim.data)

print("\nVerification:")
print(np.allclose(state_random.data, state_sim.data))
```
