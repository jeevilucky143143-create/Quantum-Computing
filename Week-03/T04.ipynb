### 2. Concept Explanation

#### Initial State
The single-qubit quantum system begins initialized in the ground computational basis state:
$$|0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$$

#### Equal Superposition via the Hadamard Gate
The single-qubit Hadamard gate ($H$) transforms deterministic computational basis states into symmetric equal-superposition states:
$$H = \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}$$
Applying the Hadamard gate to the initial ground state $|0\rangle$ produces the $|+\rangle$ state:
$$H|0\rangle = \frac{|0\rangle + |1\rangle}{\sqrt{2}} = \begin{pmatrix} \frac{1}{\sqrt{2}} \\ \frac{1}{\sqrt{2}} \end{pmatrix}$$

#### Measurement Collapse and Quantum Randomness
Unlike classical deterministic physics, quantum mechanics introduces fundamental randomness upon measurement. When a qubit in superposition $\frac{|0\rangle + |1\rangle}{\sqrt{2}}$ is measured in the standard computational basis $\{|0\rangle, |1\rangle\}$, its quantum state irreversibly collapses to either $|0\rangle$ or $|1\rangle$.

#### Theoretical Outcome Probabilities (Born Rule)
According to the **Born rule**, the probability of observing basis state $|x\rangle$ is the squared magnitude of its probability amplitude:
$$P(0) = |\langle 0 | \psi \rangle|^2 = \left|\frac{1}{\sqrt{2}}\right|^2 = \frac{1}{2} = 0.5 \quad (50\%)$$
$$P(1) = |\langle 1 | \psi \rangle|^2 = \left|\frac{1}{\sqrt{2}}\right|^2 = \frac{1}{2} = 0.5 \quad (50\%)$$

#### Finite Sampling and Statistical Fluctuations
When executing a quantum circuit for a finite number of shots ($N = 1000$), the observed frequencies will fluctuate around the theoretical mean of $50\%$. The standard deviation of the sampling error for a binomial distribution is:
$$\sigma = \sqrt{\frac{p(1-p)}{N}} = \sqrt{\frac{0.5 \times 0.5}{1000}} \approx 0.0158 \quad (1.58\%)$$
Consequently, slight variations around $0.5$ in 1000 shots are completely natural and reflect genuine statistical sampling noise.

#### Coin Toss Outcome Mapping
The measurement outcomes are mapped to classical coin-toss events:
- **Outcome `0`** $\longrightarrow$ **Heads**
- **Outcome `1`** $\longrightarrow$ **Tails**

```python
import random
import numpy as np

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator
```

### Helper Function
Define a reusable helper function `show_state(circuit, title)` to inspect statevectors and circuit drawings.

```python
def show_state(circuit, title):
    state = Statevector.from_instruction(circuit)
    print("="*60)
    print(title)
    print(circuit.draw())
    print("\nStatevector:")
    print(state)
    print("="*60)
    return state
```

---

## Create the Quantum Coin Circuit

### Apply the Hadamard Gate
Initialize a single-qubit quantum circuit (`QuantumCircuit(1)`) in the state $|0\rangle$ and apply a Hadamard gate using `coin.h(0)` to prepare an equal superposition of $|0\rangle$ and $|1\rangle$.

### Measure the Qubit
Attach measurement operations using `coin.measure_all()` to measure the qubit in the computational basis and store the result in a classical bit.

```python
coin = QuantumCircuit(1)

coin.h(0)

coin.measure_all()

print(coin.draw())
```

---

## Run the Simulation

Execute the quantum coin-toss circuit for 1000 shots using Qiskit Aer's `AerSimulator()` to sample measurement outcomes.

```python
sim = AerSimulator()

job = sim.run(coin, shots=1000)

result = job.result()

counts = result.get_counts()

print("Measurement Counts:\n")
print(counts)
```

---

## Calculate Heads and Tails Probabilities

Extract the counts for outcome `0` (Heads) and outcome `1` (Tails), normalize by the total number of shots ($N = 1000$), and compute the empirical probabilities.

```python
shots = sum(counts.values())

prob_0 = counts.get('0',0)/shots

prob_1 = counts.get('1',0)/shots

print("Probability of Heads (0):",prob_0)

print("Probability of Tails (1):",prob_1)
```

## Interpret the Results

### Comparison of Observed vs. Theoretical Outcomes

| Outcome | Bit Value | Observed Counts ($N = 1000$) | Empirical Probability | Theoretical Probability | Deviation |
| :---: | :---: | :---: | :---: | :---: | :---: |
| **Heads** | `0` | 507 | 0.507 ($50.7\%$) | 0.500 ($50.0\%$) | $+0.007$ ($+0.7\%$) |
| **Tails** | `1` | 493 | 0.493 ($49.3\%$) | 0.500 ($50.0\%$) | $-0.007$ ($-0.7\%$) |

### Statistical Analysis
- The observed distribution yields 507 Heads ($50.7\%$) and 493 Tails ($49.3\%$).
- The deviation of $\pm 0.7\%$ from the exact theoretical probability of $50\%$ falls comfortably within the expected $1\sigma$ standard sampling error ($\sigma \approx 1.58\%$).
- This confirms that the Hadamard gate acts as a balanced quantum random number generator producing true quantum coin-toss behavior.
