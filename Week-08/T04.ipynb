```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# DEUTSCH-JOZSA QUERY COMPLEXITY
# n = 2 TO 10
# ---------------------------------------------------------

n_values = np.arange(2, 11)

# ---------------------------------------------------------
# CLASSICAL WORST-CASE QUERY COUNT
# ---------------------------------------------------------
# A deterministic classical algorithm may need
# 2^(n-1) + 1 queries in the worst case.

classical_queries = (
    2 ** (n_values - 1)
) + 1

# ---------------------------------------------------------
# QUANTUM QUERY COUNT
# ---------------------------------------------------------
# Deutsch-Jozsa uses exactly ONE oracle query.

quantum_queries = np.ones(
    len(n_values)
)

# ---------------------------------------------------------
# DISPLAY QUERY COUNT TABLE
# ---------------------------------------------------------

print("\nDEUTSCH-JOZSA QUERY COMPLEXITY")
print("=" * 70)

print(
    "\nClassical worst-case queries "
    "vs Quantum queries"
)

print("-" * 70)

print(
    f"{'Input Qubits':<18}"
    f"{'Classical Worst Case':<25}"
    f"{'Quantum Queries'}"
)

print("-" * 70)

for n, classical, quantum in zip(
    n_values,
    classical_queries,
    quantum_queries
):

    print(
        f"{n:<18}"
        f"{classical:<25}"
        f"{int(quantum)}"
    )


# ---------------------------------------------------------
# CALCULATE QUERY GAP
# ---------------------------------------------------------

print("\nQUERY GAP")
print("-" * 70)

for n, classical in zip(
    n_values,
    classical_queries
):

    gap = classical / 1

    print(
        f"n = {n:<3} | "
        f"Classical = {classical:<5} | "
        f"Quantum = 1 | "
        f"Classical/Quantum = {gap:.0f}x"
    )


# ---------------------------------------------------------
# FINAL VALUES
# ---------------------------------------------------------

print("\nFINAL COMPARISON")
print("-" * 70)

print(
    f"For n = 2, classical worst case = "
    f"{classical_queries[0]} queries"
)

print(
    f"For n = 10, classical worst case = "
    f"{classical_queries[-1]} queries"
)

print(
    "Quantum query count for every n = 1"
)

print(
    "\nThe classical worst-case query count "
    "increases exponentially with n, while "
    "the Deutsch-Jozsa quantum algorithm "
    "uses one oracle query."
)


# ---------------------------------------------------------
# VISUALIZATION
# ---------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(10, 6)
)

ax.plot(
    n_values,
    classical_queries,
    marker="o",
    linewidth=2,
    label="Classical Worst Case"
)

ax.plot(
    n_values,
    quantum_queries,
    marker="s",
    linewidth=2,
    label="Quantum - Deutsch-Jozsa"
)

ax.set_title(
    "Classical vs Quantum Query Count"
)

ax.set_xlabel(
    "Number of Input Qubits (n)"
)

ax.set_ylabel(
    "Number of Oracle Queries"
)

ax.set_xticks(
    n_values
)

# Log scale makes the exponential growth
# clearly visible.

ax.set_yscale(
    "log"
)

ax.grid(
    True,
    alpha=0.3
)

ax.legend()

plt.tight_layout()

plt.show()
```
